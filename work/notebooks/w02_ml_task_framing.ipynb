I chose scoring because the output is not only a yes/no decision. A ranked score lets us compare many pages and decide which ones should be reviewed first.

## 3. Success metric

### My answer

My primary success metric is **Precision@K**.

I will use the model to rank content pages by their predicted refresh opportunity score. Precision@K measures how many of the top K pages actually show the defined future performance decline.

A higher Precision@K means the ranked queue contains more pages that later match the outcome we wanted to identify.

This metric fits the real decision because the content team cannot review every page. They need a short, prioritized list of pages to review first.

## 5. Why ML beats a fixed rule here

### My answer

A fixed rule would use one or two thresholds, such as refreshing every page older than a certain number of days. This can miss important differences between pages.

Refresh opportunity can depend on several signals together, such as impressions, clicks, CTR, average position, sessions, content age, and changes in performance.

ML can learn combinations of these signals and produce a ranked score instead of applying the same rule to every page.

The output is intended as **decision support**: it helps prioritize which pages should be reviewed first. It does not prove that refreshing a page will cause better performance.
