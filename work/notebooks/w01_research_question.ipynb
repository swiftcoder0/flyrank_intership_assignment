## 2. The question: decision, action, cost of a wrong call

**Research question:** Which pages should be prioritized for content review or refresh based on observable signals such as impressions, CTR, content age, average position, and time since the last update?

**Unit of analysis:** One webpage.

**Output:** A ranked list or priority score showing which pages may deserve review first.

**Decision:** The work should help decide which pages should receive attention first when the available time or resources are limited.

**Action:** A content or SEO team could review the highest-priority pages and then decide whether each page needs an update, optimization, or no change.

**Cost of a wrong call:** A false positive could waste time by sending a healthy page for unnecessary review. A false negative could be more costly because a page that genuinely needs attention might be missed and continue underperforming.

**Why data/ML can help:** Several signals may matter at the same time, and their relationships may not be captured well by one fixed rule. Data analysis and ML can help combine these signals into a more useful prioritization system, while the final content decision remains with a human reviewer.

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains **30,000 pages**. Of these, **16,262 pages (54.2%)** are marked with a downward trend, which shows that declining pages are common enough in this dataset to make prioritization a meaningful problem.

The median time since the last update is only **20 days**. This suggests that simply treating older or stale pages as the highest priority may not be sufficient. A useful refresh-prioritization approach may need to consider multiple observable signals, such as impressions, CTR, content age, average position, and update history.

These observations make the Refresh / Content Opportunity Scoring lane worth investigating further, while they do not by themselves show that refreshing a page will improve its performance.


```python
import os
import subprocess

REPO_URL = "https://github.com/swiftcoder0/flyrank_intership_assignment.git"
REPO_DIR = "flyrank_intership_assignment"

# Clone repo if it is not already available
if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", REPO_URL, REPO_DIR], check=True)

os.chdir(REPO_DIR)

print("Working directory:", os.getcwd())

print("Dataset exists:", os.path.exists("data/raw/content_refresh_anonymized.csv"))
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)

declining_pages = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .sum()
)

declining_pct = declining_pages / total_pages * 100
median_days_update = df["days_since_last_update"].median()

print(f"Total pages: {total_pages:,}")
print(f"Declining pages: {declining_pages:,} ({declining_pct:.1f}%)")
print(f"Median days since last update: {median_days_update:.0f}")
```

## 4. Careful words: what I can and can't claim

My work can identify **observed patterns and associations** between page-level signals such as impressions, CTR, average position, content age, update history, and declining performance. It may provide **decision-support** by helping prioritize pages for human review.

However, I cannot claim that these relationships are causal. A high priority score would not prove that a page needs to be refreshed or that refreshing it will improve traffic or rankings. I also cannot claim that the model predicts Google's algorithm. The results should be treated as directional evidence that can help a human decide what to investigate next.
